
# Visualization of generated data

```python
from pathlib import Path

import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap

from synthbold.config import Config
from synthbold.geometries import Toroids
from synthbold.io import save_mesh, save_nifti
from synthbold.visualize import make_gif, render_3d

out_dir = Path("./visualize")
out_dir.mkdir(exist_ok=True, parents=True)

# load default configuration
config = Config()

# Get tab20 colormap and prepend black for 0 for vessel visualization (each vessel gets
# a different color)
base_cmap = plt.colormaps["tab20"].resampled(20)
colors = ["black"] + [base_cmap(i) for i in range(1, 20)]
cmap_vessel = ListedColormap(colors)
```

```python
# initialize generator
generator = Toroids.from_config(config)
data = generator(n_sample=10)

# save batch as gif
make_gif(out_dir / "toroids.gif", data, cmap=cmap_vessel, mip=True)

# save first batch entry as 3D render
render_3d(out_dir / "toroids_render.png", data[0])

# save batch as nifti file to disk
data_numpy = data.cpu().numpy()
save_nifti(out_dir / "toroids.nii", data_numpy, permute=True)

# save first batch entry (volume) as mesh file
data_binary = data_numpy[0] != 0
save_mesh(out_dir / "toroids_label_mesh", data_binary)
```
